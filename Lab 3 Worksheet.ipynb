```python
from bs4 import BeautifulSoup
import re

with open('bongsoon.html', 'r', encoding='utf-8') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, 'html.parser')
```

Check that the parsing was successful:

```python
soup.title
```

## exercise 1 

```python
#1
meta = soup.find("meta", attrs={"name": "description"})
content = meta['content']
```

```python
#1A

year =re.findall(r'\b\d{4}\b', content)
print(year)

#1b
letters = re.findall(r'\b[a-zA-Z]{6}\b', content)
print(letters)

#1c 
print(re.findall(r'\b\d{4}\b', content))
print(re.findall(r'\b[a-zA-Z]{6}\b', content))
```

### Exercise 2

```python
text = soup.find('div', class_="show-synopsis").find('p')
text1 = text.get_text(strip=True)
print(text)
```

```python
#2a
vowels = re.findall(r'\b[aeiou]\w*\b', text1)
print(vowels)
```

```python
#2b
filter1 = re.findall(r'\b[aeiou]\w*b', text1)
print(filter1)
```
